```python
A = torch.randn(3, 4)
B = torch.randn(4, 5)


print(torch.round(A @ B, decimals=2))
```

```python
print(np.round(A@B, 2))  # np.round works for tensor
```

```python
C1 = np.random.randn(4, 7)
print(np.round(A.numpy()@C1, 2))
```

```python
C2 = torch.tensor(C1, dtype=torch.float)
# C1 is float64/double, for performing matmul with A which is float32/float, we need to typecast
print(np.round(A@C2, 2))
```

```python
A = np.random.randn(3, 4)
B = np.random.randn(4, 5)
print(np.round(A@B, 2))
# torch.round doesn't works on ndarray, we need to convert to tensor
print(torch.round(torch.tensor(A@B), decimals=2))
```

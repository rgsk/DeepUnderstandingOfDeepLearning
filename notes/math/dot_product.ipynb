```python
import numpy as np
import torch
```

```python
# create a vector
nv1 = np.array([1, 2, 3, 4])
nv2 = np.array([0, 1, 0, -1])

# dot product via function
print(np.dot(nv1, nv2))
print(nv1 * nv2)
# dot product via computation
print(np.sum(nv1*nv2))
```

```python
# Dot product in 2D

nv1 = np.array([[0, 3, 2], [-3, -3, 1], [1, 0, 2]])
nv2 = np.array([[1, 0, 6], [2, -1, 0], [5, 1, 4]])

np.dot(nv1, nv2)  # matrix multiplication
```

```python
nv1 * nv2
```

```python
np.sum(nv1 * nv2)
```

```python
# create a vector
tv1 = torch.tensor([1, 2, 3, 4])
tv2 = torch.tensor([0, 1, 0, -1])

# dot product via function
print(torch.dot(tv1, tv2))

# dot product via computation
print(torch.sum(tv1*tv2))
```

```python
tv1 = torch.tensor([[0, 3, 2], [-3, -3, 1], [1, 0, 2]])
tv2 = torch.tensor([[1, 0, 6], [2, -1, 0], [5, 1, 4]])

try:
    print(torch.dot(tv1, tv2))
except Exception as e:
    print(f'Error: {e}')
```

```python
tv1*tv2
```

```python
print(torch.sum(tv1*tv2))
```
